### Limitations

Every analysis has them, and naming them is a strength rather than an admission. Be concrete:

- **What the data cannot tell you.** A listing's price is not what it was booked at. A snapshot is not a time series.
- **Quality problems that survived the cleaning.** The column that was 40% null and the rows you dropped.
- **Correlation, not causation.** Two things moving together is where your analysis ends.
- **Coverage.** One city, two years, one platform. Say what you cannot generalise to.
